# J-Quants V2 原データ取得

CLI と同じ `stock_prediction.data_download` を使い、株価日足と閉場日を含む取引カレンダーを `data/raw/` の新規スナップショットへ保存します。APIキーは非表示入力し、Notebookやmanifestには保存しません。開始日・終了日は契約プランで取得できる期間内に設定してください。

```python
from getpass import getpass
from pathlib import Path
from stock_prediction.data_download import DownloadConfig, download_jquants_raw

root = next(
    path for path in (Path.cwd(), *Path.cwd().parents)
    if (path / "README.md").is_file() and (path / "src" / "stock_prediction").is_dir()
)
```

```python
start_date = "2026-10-01"
end_date = "2026-10-07"
requests_per_minute = 5  # 契約プランの上限に合わせる（5はFreeプラン向けの保守値）
api_key = getpass("J-Quants API key: ")
```

```python
config = DownloadConfig(
    start_date=start_date,
    end_date=end_date,
    output_root=root / "data" / "raw",
    requests_per_minute=requests_per_minute,
)
result = download_jquants_raw(config, api_key=api_key)
result.output_dir
```

```python
{
    "status": result.manifest["status"],
    "calendar_rows": result.manifest["calendar"]["rows"],
    "trading_dates": result.manifest["totals"]["completed_trading_dates"],
    "price_rows": result.manifest["totals"]["price_rows"],
}
```

長期間の取得前に、まず数営業日の範囲で契約・レート設定・出力を確認してください。失敗または中断時も同じスナップショット内の `manifest.json` と `partial_prices/`（取得済み行がある場合）を確認できます。同じ出力先は再利用せず、新しい実行で再取得します。
